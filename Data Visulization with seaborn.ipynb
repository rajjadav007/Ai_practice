```python
import seaborn as sns
```

```python
## basic Plotting with seaborn

tips=sns.load_dataset('tips')
tips
```

```python
## create scatter plot 
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("scatter plot of total bill vs tip")
plt.show()
```

```python
##line plot 

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("line plot title")
plt
```

```python
## catagorical plot
## bar plot

sns.barplot(x='day',y='total_bill',data=tips)
plt.title("bar plot of title bill by day")
plt
```

```python
## box plot

sns.boxenplot(x='day',y='total_bill',data=tips)
plt
```

```python
## violin plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
## histograms

sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
## kde plot 

sns.kdeplot(tips['total_bill'],shade=True)
```

```python
## commnly used pairplot

sns.pairplot(tips)
```

```python
tips
```

```python
## heatmap

corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd
sales_df=pd.read_csv('100 Sales Records.csv')
sales_df.head()
```

```python
## plot total sales by product
plt.Figure(figsize=(10,6))
sns.barplot(x='Item Type',y='Total Revenue',data=sales_df,estimator=sum)
plt.xlabel("item name")
plt.ylabel("total ravenue")
plt.title("total sales")
plt.show()
```

```python
## plot total sales by rigion
plt.Figure(figsize=(10,6))
sns.barplot(x='Region',y='Total Revenue',data=sales_df,estimator=sum)
plt.xlabel("item name")
plt.ylabel("total ravenue")
plt.title("total Region")
plt.show()
```
