```python
## ranaming columns
df=df.rename(columns={"minute":"minute_time"})

df.head()
```

```python
#change datatype
df["player_number"]=df["player_id"].astype(float)

df.head()
```

```python
#data aggreating and grouping

grouped_mean=df.groupby('event_id')['player_id'].mean()
print (grouped_mean)
```

```python
grouped_sum=df.groupby(["event_id","event_type"])["minute_time"].sum()
print (grouped_sum)
```

```python
df.groupby(["event_id","event_type"])["minute_time"].mean()
```

```python
##multipal aggregate function
groouped_age=df.groupby("event_id")["team_id"].agg(['mean','sum','count'])
groouped_age
```

```python
## meargin and joinning dataframes
df1=pd.DataFrame({'Key': ["a", "b",'c'], "Value1": [1,2,3]})
df2=pd.DataFrame({'Key': ["d", "e","f"], "Value2": [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
##merge dataframe on the key collum
pd.merge(df1,df2,on="Key",how="inner")
```

```python
pd.merge(df1,df2,on="Key",how="outer")
```

```python
pd.merge(df1,df2,on="Key",how="right")
```

```python
pd.merge(df1,df2,on="Key",how="left")
```
